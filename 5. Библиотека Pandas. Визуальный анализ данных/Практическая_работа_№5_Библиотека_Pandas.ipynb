6. Постройте сводную таблицу с помощью метода pivot_table(),

Примерная структура таблицы (в качестве значений выводится количество людей, относящихся к той или иной группе):

image.png

````python
summary_table = pd.DataFrame()
summary_table['education'] = df['education']
summary_table['capital.diff'] = df['capital.gain'] - df['capital.loss']

summary_table = summary_table[summary_table['capital.diff'] != 0]

summary_table['categories'] = pd.qcut(summary_table["capital.diff"], q=10, duplicates='drop')

pivot_cap = summary_table.pivot_table(index='education',
                                      columns='categories',
                                      values='capital.diff',
                                      aggfunc='count',
                                      fill_value=0)
pivot_cap
````

15) Создайте случайную колонку - magic_salary, которую нужно будет вычислить следующим образом: если зарплата небольшая (<50K), тогда случайно выберите число из диапазона [0,50]. Если зарплата выше 50K тогда из диапазона [51, 200]. Посчитайте среднюю зарплату в час для групп людей с одни уровнем образования на основе нашей случайной колонки magic_salary

````python
import numpy as np
def generate_magic_salary(income):
    if income == '<=50K':
        return np.random.randint(0, 51)
    else:
        return np.random.randint(51, 201)
df['magic_salary'] = df['income'].apply(generate_magic_salary)
df['hourly_magic_rate'] = df['magic_salary'] / df['hours.per.week']
avg_hourly_by_edu = df.groupby('education')['hourly_magic_rate'].mean().sort_values(ascending=False)
print(avg_hourly_by_edu)
````
